Test differentiable simulation and visualization packages.

```python
import nimblephysics as nimble  # pip install nimblephysics
import torch

# Load your aircraft model (convert URDF → .skel) into the DART world
world = nimble.loadWorld("aircraft.skel")

# Initialize state and control as PyTorch tensors (on GPU if available)
state   = torch.zeros(world.getStateSize(), device="cuda")
control = torch.zeros(world.getActionSize(), device="cuda")

# Simulate for T timesteps
states = []
for _ in range(100):
    state = nimble.timestep(world, state, control)  # differentiable step
    states.append(state)

# Visualize rollout in an interactive 3D GUI
gui = nimble.NimbleGUI(world)
gui.serve(8080)          # host at http://localhost:8080
gui.loopStates(states)   # animate
gui.blockWhileServing()  # block until closed
```
